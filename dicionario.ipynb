### **Variáveis não utilizadas**



**Data leakage** (`APGAR1`, `APGAR5`, `IDANOMAL`, `CODANOMAL`, `STCESPARTO`, `TPROBSON`): Essas variáveis foram excluidas por representarem informação disponível apenas após o desfecho que o modelo busca prever. Anomalia congênita e métricas de Apgar são avaliadas após o momento de nascimento e compartilham a mesma causa fisiologica de baixo peso, o que também gera uma correlaçaõ artificial e infla métricas de validação sem produzir um modelo utilizável em um cenário real de triagm pré-parto. Variáveis como `STCESPARTO` e `TPROBSON` seguem a mesma lógica por proximidade causal, o primeiro é uma decisão tomada no momento do parto, e o segundo é um índice derivado que já incorpora `PARTO`, `GRAVIDEZ` e outras variáveis do modelo, criando redundância.


**Variáveis sensiveis** (`RACACOR`, `RACACORMAE`): raça/cor foi removida por ter um papel ambíguo, correlaciona desigualdade de acesso a saúde, usa-la como preditora para um evento de desfecho clínico levanta viés discriminatorio caso o modelo seja utilizado em produção.

**Zona de causalidade invertida** (`PARTO`, `STTRABPART`, `TPAPRESENT`): esses campos são conhecidos antes e depois do parto, mas a decisão por cesárea ou indução, pode ser motivada por um risco de baixo peso ou sofrimento fetal já identificado no pré-natal, e não o contrário.

**Metadados administrativos e de processamento** (`ORIGEM`, `CODESTAB`, `LOCNASC`, `CODOCUPMAE`, `HORANASC`, `DTCADASTRO`, `NUMEROLOTE`, `VERSAOSIST`, `DTRECEBIM`, `DIFDATA`, `DTRECORIGA`, `NATURALMAE`, `CODMUNNATU`, `CODUFNATU`, `SERIESCMAE`, `DTNASCMAE`, `IDADEPAI`, `DTULTMENST`, `TPNASCASSI`, `TPFUNCRESP`, `TPDOCRESP`, `DTDECLARAC`, `ESCMAEAGR1`, `STDNEPIDEM`, `STDNNOVA`, `CODPAISRES`, `CONTADOR`): excluídos em bloco por não carregarem informação causal sobre o evento clínico do nascimento.

**Variáveis substituídas por versão mais informativa** (`ESCMAE`, `GESTACAO`, `CONSULTAS`): removidas por serem versões categorizadas em faixas de campos que possuem equivalentes mais granulares no proprio dataset.
