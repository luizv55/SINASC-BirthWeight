## Dicionário de Dados - SINASC e SIM

#### SINASC - Sistema de Informações sobre Nascidos Vivos

##### Identificação e local
| Campo | Descrição | Domínio |
| --- | --- | --- |
| CODMUNASC | Município de ocorrência do nascimento | Código IBGE |
| CODMUNRES | Município de residência da mãe | Código IBGE |
| LOCNASC | Local do nascimento | 1=Hospital, 2=Outro estab. saúde, 3=Domicilio, 4=Outros |
| CODESTAB | Estabelecimento de saúde(CNES) | Código CNES |
| DTNASC | Data de nascimento | DDMMAAAA |
| HORANASC | Hora de nascimento | HHMM |

##### Dados da mãe

| Campo | Descrição | Domínio |
| --- | --- | --- |
| IDADEMAE | Idade da mãe em anos | Numérico |
| ESTCIVMAE | Estado civil da mãe | 1=Solteira, 2=Casada, 3=Viúva, 4=Separada, 5=União estável, 9=Ignorado |
| ESCMAE | Escolaridade da mãe (anos de estudo) | 1=Nenhuma, 2=1-3 anos, 3=4-7anos, 4=8-11 anos, 5=12+anos |
| CODOCUPMAE | Ocupação da mãe | Código CBO |
| RACACORMAE | Raça/cor da mãe | 1=Branca, 2=Preta, 3=Amarela, 4=Parda, 5=Indígena |
| QTDFILVIVO | Nº de filhos vivos | Numérico |
| QTDFILMORT | Nº de filhos mortos | Numérico |




##### Gestação e parto
| Campo | Descrição | Domínio |
| --- | --- | --- |
| GESTACAO | Semanas de gestação (Faixa) | 1=<22, 2=22–27, 3=28–31, 4=32–36, 5=37–41, 6=42+ |
| SEMAGESTAC | Semanas de gestação (número exato) | Numérico |
| GRAVIDEZ | Tipo de gravidez | 1=Única, 2=Dupla, 3=Tripla+ |
| PARTO | Tipo de parto | 1=Vaginal, 2=Cesáreo |
| CONSULTAS | Nº de consultas de pré-natal (faixa) | 1=Nenhuma, 2=1-3, 3=4-6, 4=7+ |


##### Recém-nascido
| Campo | Descrição | Domínio |
| --- | --- | --- |
| SEXO | Sexo do RN | 0=Igorado, 1=Masculino, 2=Feminino |
| PESO | Peso ao nascer (gramas) | Numérico |
| APGAR1 | Apagr no 1º minuto | 0-10 |
| APGAR5 | Apagr no 5º minuto | 0-10 |
| RACACOR | Raça/cor do RN | Mesmo domínio de RACACORMAE |
| IDANORMAL | Anomalia congênita identificada | 1=Sim, 2=Não |
| CODANORMAL | Código CID-10 da analomalia | Ex:Q00-Q99 |

### **Variáveis não utilizadas**



**Data leakage** `APGAR1`, `APGAR5`, `IDANOMAL`, `CODANOMAL`, `STCESPARTO`, `TPROBSON`: Essas variáveis foram excluidas por representarem informação disponível apenas após o desfecho que o modelo busca prever. Anomalia congênita e métricas de Apgar são avaliadas após o momento de nascimento e compartilham a mesma causa fisiologica de baixo peso
